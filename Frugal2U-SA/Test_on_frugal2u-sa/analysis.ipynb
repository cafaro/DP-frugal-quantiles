```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
plt.rc('lines', linewidth=0.5, color='k')
font = {'family' : 'sans-serif',
        'weight' : 'regular',
        'size'   : '12'}
plt.rc('font', **font)
plt.rcParams['savefig.bbox']='tight'
dist_name = ['normal', 'cauchy', 'uniform', 'exp', 'chisquared', 'gamma', 'lognorm', 'extreme']
```

```python
def plot_chunks(ax):
    df = pd.read_csv('test_on_chunks.csv')
    dx = df['chunks']
    dy = df['nae']
    el = df['nae_cil']
    er = df['nae_cir']
    ci = [el.to_numpy(),er.to_numpy()]
    
    # ax.set_title(r'Error, quantile: 0.99 length: 10M distribution: Normal $\epsilon$:1', fontsize=18, style='italic')
    ax.set_ylim([0,3])
    ax.set_xticks([2, 4, 8, 16])
    # ax.xaxis.set_tick_params(width=0.5)
    # ax.yaxis.set_tick_params(width=0.5)
    r = ax.errorbar(dx, dy, yerr=ci, markersize=9, linewidth=0.5, elinewidth=0.3, capthick=0.3,  capsize=3, label='frugal2u-sa')
    ax.get_lines()[0].set_color("black")
    r[-1][0].set_linestyle('-.')
    r[-1][0].set_color("black")
    ax.set_xlabel('chunks', fontsize=34)
    ax.set_ylabel('Normalized Abs Error', fontsize=34)
    [x.set_linewidth(0.5) for x in ax.spines.values()]
    ax.tick_params(axis='both', which='major', labelsize=30)
    
    return ax

def plot_eps(ax: plt.Axes):
    df = pd.read_csv('test_on_epsilon.csv')
    dx = df['epsilon']
    dy = df['nae']
    el = df['nae_cil']
    er = df['nae_cir']
    ci = [el.to_numpy(),er.to_numpy()]
    
    # ax.set_title(r'Error, quantile: 0.99 length: 10M distribution: Normal chunks: 4', fontsize=18, style='italic')
    ax.set_ylim([0,3])
    ax.set_xticks([0.1, 0.5, 1, 2])
    # ax.xaxis.set_tick_params(width=0.5)
    # ax.yaxis.set_tick_params(width=0.5)
    r = ax.errorbar(dx, dy, yerr=ci, markersize=9, linewidth=0.5, elinewidth=0.3, capthick=0.3,  capsize=3, label='frugal2u-sa')
    ax.get_lines()[0].set_color("black")
    r[-1][0].set_linestyle('-.')
    r[-1][0].set_color("black")
    ax.set_xlabel(r'$\epsilon$', fontsize=34)
    ax.set_ylabel('Normalized Abs Error', fontsize=34)
    [x.set_linewidth(0.5) for x in ax.spines.values()]
    ax.tick_params(axis='both', which='major', labelsize=30)
    
    return ax

def plot_quantile(ax):
    df = pd.read_csv('test_on_quantile.csv')
    dx = df['q']
    dy = df['nae']
    el = df['nae_cil']
    er = df['nae_cir']
    ci = [el.to_numpy(),er.to_numpy()]
    
    # ax.set_title(r'Error, length: 10M distribution: Normal chunks: 4 $\epsilon$: 1', fontsize=18, style='italic')
    ax.set_ylim([0,3])
    ax.set_xticks([0.1, 0.3,  0.5, 0.99])

    # ax.xaxis.set_tick_params(width=0.5)
    # ax.yaxis.set_tick_params(width=0.5)
    r = ax.errorbar(dx, dy, yerr=ci, markersize=9, linewidth=0.5, elinewidth=0.3, capthick=0.3,  capsize=3, label='frugal2u-sa')
    ax.get_lines()[0].set_color("black")
    r[-1][0].set_linestyle('-.')
    r[-1][0].set_color("black")
    ax.set_xlabel(r'quantile', fontsize=34)
    ax.set_ylabel('Normalized Abs Error', fontsize=34)
    [x.set_linewidth(0.5) for x in ax.spines.values()]
    ax.tick_params(axis='both', which='major', labelsize=30)
    
    return ax

def plot_length(ax):
    df = pd.read_csv('test_on_length.csv')
    dx = df['n']
    dy = df['nae']
    el = df['nae_cil']
    er = df['nae_cir']
    ci = [el.to_numpy(),er.to_numpy()]
    
    # ax.set_title(r'Error, quantile: 0.99 distribution: Normal chunks: 4 $\epsilon$: 1', fontsize=18, style='italic')
    ax.set_ylim([0,3])
    ax.set_xticks([10000000, 50000000, 75000000, 100000000])
    # ax.xaxis.set_tick_params(width=0.5)
    # ax.yaxis.set_tick_params(width=0.5)
    r = ax.errorbar(dx, dy, yerr=ci, markersize=9, linewidth=0.5, elinewidth=0.3, capthick=0.3,  capsize=3, label='frugal2u-sa')
    ax.get_lines()[0].set_color("black")
    r[-1][0].set_linestyle('-.')
    r[-1][0].set_color("black")
    ax.set_xlabel(r'stream length', fontsize=34)
    ax.set_ylabel('Normalized Abs Error', fontsize=34)
    [x.set_linewidth(0.5) for x in ax.spines.values()]
    ax.tick_params(axis='both', which='major', labelsize=30)
    
    return ax

def plot_distro(ax: plt.Axes):
    df = pd.read_csv('test_on_distribution.csv')
    dx = [dist_name[i] for i in df['distr'].map(int)]
    dy = df['nae']
    el = df['nae_cil']
    er = df['nae_cir']
    ci = [el.to_numpy(),er.to_numpy()]
    
    # ax.set_title(r'Error, quantile: 0.99 length: 10M chunks: 4 $\epsilon$: 1', fontsize=18, style='italic')
    ax.set_ylim([0,1.0])
    # ax.xaxis.set_tick_params(width=0.5)
    # ax.yaxis.set_tick_params(width=0.5)
    ax.bar(x=dx, height=dy, color='gray')
    r = ax.errorbar(dx, dy, fmt='none', yerr=ci, markersize=9, linewidth=0.5, elinewidth=0.3, capthick=0.3,  capsize=3, label='frugal2u-sa')
    # ax.get_lines()[0].set_color("black")
    r[-1][0].set_linestyle('-.')
    r[-1][0].set_color("black")
    ax.set_xlabel(r'', fontsize=34)
    ax.set_ylabel('Normalized Abs Error', fontsize=20)
    [x.set_linewidth(0.5) for x in ax.spines.values()]
    ax.tick_params(axis='both', which='major', labelsize=15)
    return ax

def plot_distro_box(ax: plt.Axes):
    dfs = [pd.read_csv(f'test_on_distribution_{i}.csv') for i in range (1, 9)]
    nae_series = [dfs[i]['nae'].to_list() for i in range(0, 8)]
   
    ax.boxplot(nae_series, patch_artist=True, tick_labels=dist_name, 
                showmeans=False, showfliers=False,
                medianprops={"color": "black", "linewidth": 0.5},
                boxprops={"facecolor": "lightgray", "edgecolor": "black",
                            "linewidth": 0.5},
                whiskerprops={"color": "gray", "linewidth": 1.0, "linestyle": "dashed"},
                capprops={"color": "gray", "linewidth": 1.0, "linestyle": "dashed"}) 
    ax.set_xlabel(r'', fontsize=34)
    ax.set_ylabel('Normalized Abs Error', fontsize=20)
    [x.set_linewidth(0.5) for x in ax.spines.values()]
    ax.tick_params(axis='both', which='major', labelsize=15)
    return ax
```

```python
fig, ax = plt.subplots(1,1, figsize=(12,8))
plot_chunks(ax)
fig.savefig(f'dperror_chunks_2u.eps', dpi=600, bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots(1,1, figsize=(12,8))
plot_length(ax)
fig.savefig(f'dperror_n_2u.eps', dpi=600, bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots(1,1, figsize=(12,8))
plot_quantile(ax)
fig.savefig(f'dperror_quantile_2u.eps', dpi=600, bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots(1,1, figsize=(12,8))
plot_eps(ax)
fig.savefig(f'dperror_epsilon_2u.eps', dpi=600, bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots(1,1, figsize=(12,8))

plot_distro_box(ax)
fig.savefig(f'dperror_distrib_2u.eps', dpi=600, bbox_inches='tight')
plt.show()
```
